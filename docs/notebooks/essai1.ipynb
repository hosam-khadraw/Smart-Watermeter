```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import FastICA

# Créer des signaux sources indépendants
np.random.seed(0)
n_samples = 2000
time = np.linspace(0, 8, n_samples)

# Trois signaux sources
S1 = np.sin(2 * time)  # Signal sinusoïdal
S2 = np.sign(np.sin(3 * time))  # Signal carré
S3 = np.random.rand(n_samples)  # Bruit aléatoire

# Combinaison linéaire des signaux sources
S = np.c_[S1, S2, S3]
S /= S.std(axis=0)  # Standardisation

# Mélanger les signaux
A = np.array([[1, 1, 1], [0.5, 2, 1.0], [1.5, 1.0, 2.0]])  # Matrice de mélange
X = np.dot(S, A.T)  # Signaux mélangés

# Appliquer ICA
ica = FastICA(n_components=3)
S_ = ica.fit_transform(X)  # Signal source estimé
A_ = ica.mixing_  # Matrice de mélange estimée

# Vérification de la séparation
assert np.allclose(X, np.dot(S_, A_.T) + ica.mean_)

# Visualisation
plt.figure(figsize=(9, 7))

plt.subplot(3, 1, 1)
plt.title("Signaux Mélangés")
plt.plot(X)

plt.subplot(3, 1, 2)
plt.title("Signaux Séparés (par ICA)")
plt.plot(S_)

plt.subplot(3, 1, 3)
plt.title("Signaux Sources Originales")
plt.plot(S)

plt.tight_layout()
plt.show()
```
